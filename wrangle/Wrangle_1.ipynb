## EPOCH_SECONDS

### 1. Check transaction times are between 0 and today

```python
# Get current Unix timestamp (seconds)
current_time = int(time.time())

# Find the longest/latest transaction timestamp
latest_transaction = df['EPOCH_SECONDS'].max()

# Verify
if latest_transaction <= current_time:
    print("PASS: All transactions are not later than today.")
else:
    print("FAIL: Some transactions have timestamps later than today.")
```

```python
earliest_transaction_time = df['EPOCH_SECONDS'].min()

# Verify
print(
    "PASS: All time entries are valid" 
    if earliest_transaction_time >= 0 else 
    "FAIL: Some transactions are too early"
)
```

## CURRENCY_CODE

```python
# Get unique currency codes and convert to uppercase
currency_codes = df['CURRENCY_CODE'].dropna().astype(str).str.strip().str.upper().unique()

# Get valid ISO 4217 currency codes
valid_codes = {currency.alpha_3 for currency in pycountry.currencies}

# Identify invalid currency codes
invalid_codes = sorted(set(currency_codes) - valid_codes)

# Display results
if len(invalid_codes) == 0:
    print("PASS: All currency codes are valid ISO 4217 codes.")
else:
    print("FAIL: Invalid currency codes found:")
    print(invalid_codes)
```
